```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/Dataset_.csv', encoding='utf-8-sig')
print(f'Shape: {df.shape}')
df.head()
```

```python
# Missing values
df.isnull().sum()[df.isnull().sum() > 0]
```

```python
# Rating distribution
df['Aggregate rating'].value_counts().sort_index().plot(kind='bar', figsize=(12,4))
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Top cuisines
from collections import Counter
cc = Counter()
for row in df['Cuisines'].dropna():
    for c in row.split(','):
        cc[c.strip()] += 1
print('Top 15 cuisines:')
for c,n in cc.most_common(15):
    print(f'  {c}: {n}')
```
